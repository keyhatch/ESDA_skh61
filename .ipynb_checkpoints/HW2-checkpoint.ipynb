$E[ax+b] \\= \sum(ax+b)*p(x)\\= \sum(ax)*p(x) + \sum b*p(x)\\= a\sum(x)*p(x) + b \sum p(x)\\=aE[x] + b$

$Var(x) = E[(x-E(x))^2]\\=E[(x-E(x))*(x-E(x))]\\=E[x^2-xE(X)-xE(x)+E(x)^2]\\=E[x^2-2xE(x)+E(x)^2]\\=E(x^2)-E(2xE(x))+E(E(x)^2)\\=E(x^2)-2E(x)E(x)+E(x)^2\\=E(x^2)-E(x)^2$

# 2. Sentinel-2 data over Durham (20 points)

The objective of this problem is to learn how to download, process, and plot spatial data. More specifically, we are going to use data from the version 2.0 of the harmonized-landsat-sentinel-2 (HLS, https://lpdaac.usgs.gov/data/get-started-data/collection-overview/missions/harmonized-landsat-sentinel-2-hls-overview/
) dataset. Begin by opening an account on earthdata.nasa.gov. Then proceed to find and download HLS data to your Docker container for Durham, NC on 09/10/2021. More specifically, the desired bounding box is between longitudes -79.0 and -78.8 and latitudes 35.90 and 36.10. You will need to download a larger region and then subset it on the docker container. You will want to first reproject all your downloaded files to a Plate Carree projection (i.e., regular lat/lon grid) at a 1 arcsec spatial resolution. Proceed to then create a RGB plot using the corresponding Sentinel-2 bands. Converting the radiances "felt" by the satellite to RGB values is not trivial but we will keep it simple; for this problem simply normalize each corresponding band to 0-1 or 0-255 (either will work). Once this is complete, create a map of NDVI (https://gisgeography.com/ndvi-normalized-difference-vegetation-index/) as well.


```python
%%bash 
#Plate Carree projection
gdalwarp \
    -t_srs EPSG:4326 \
    -tr 0.0002777778 0.0002777778 \
    -te_srs EPSG:4326 \
    -te -79.0 35.90 -78.8 36.10 \
    HLS.S30.T17SPV.2021253T155819.v2.0.B04.tif \
    B04_Durham.tif
gdalwarp \
    -t_srs EPSG:4326 \
    -tr 0.0002777778 0.0002777778 \
    -te_srs EPSG:4326 \
    -te -79.0 35.90 -78.8 36.10 \
    HLS.S30.T17SPV.2021253T155819.v2.0.B03.tif \
    B03_Durham.tif

gdalwarp \
    -t_srs EPSG:4326 \
    -tr 0.0002777778 0.0002777778 \
    -te_srs EPSG:4326 \
    -te -79.0 35.90 -78.8 36.10 \
    HLS.S30.T17SPV.2021253T155819.v2.0.B02.tif \
    B02_Durham.tif

gdalwarp \
    -t_srs EPSG:4326 \
    -tr 0.0002777778 0.0002777778 \
    -te_srs EPSG:4326 \
    -te -79.0 35.90 -78.8 36.10 \
    HLS.S30.T17SPV.2021253T155819.v2.0.B8A.tif \
    B8A_Durham.tif
```

```python
import rasterio
import numpy as np
import matplotlib.pyplot as plt

# Read bands
with rasterio.open("B04_Durham.tif") as src:
    red = src.read(1).astype(float)

with rasterio.open("B03_Durham.tif") as src:
    green = src.read(1).astype(float)

with rasterio.open("B02_Durham.tif") as src:
    blue = src.read(1).astype(float)

with rasterio.open("B8A_Durham.tif") as src:
    nir = src.read(1).astype(float)
    
def normalize_bands(band):
    return (band - np.nanmin(band)) / (np.nanmax(band) - np.nanmin(band))

red_norm = normalize_bands(red)
green_norm = normalize_bands(green)
blue_norm = normalize_bands(blue)

rgb = np.dstack((red_norm, green_norm, blue_norm))

extent = [-79.0, -78.8, 35.90, 36.10]

plt.figure(figsize=(8, 8))
plt.imshow(rgb, extent=extent, origin="upper")
plt.xlabel("Longitude")
plt.ylabel("Latitude")
plt.title("HLS Sentinel-2 RGB — Durham, NC — 09/10/2021")
plt.show()

ndvi = (nir - red) / (nir + red)

plt.figure(figsize=(10, 10))
img = plt.imshow(
    ndvi,
    extent=extent,
    origin="upper",
    cmap="RdYlGn",
    vmin=-1,
    vmax=1)
plt.colorbar(img, label="NDVI")
plt.xlabel("Longitude")
plt.ylabel("Latitude")
plt.title("HLS Sentinel-2 NDVI — Durham, NC — 09/10/2021")
plt.show()
```

# 3. Calculating the correct spatial mean (20 pts)

A simple spatial mean of a regular lat/long grid (Plate Carree) can lead to large discrepancies in the estimated spatial mean. To assess this impact, we are going to use the variable `t2m` in the annual ERA-Interim database (`/data/era-interim/era_interim_monthly_197901_201512_upscaled.nc_ann`). Start by assembling an array of actual area per pixel and then compute for each year the area weighted mean. Then compare the resulting time series to that resulting from using the equally weighted spatial mean (i.e., naive approach). Provide an explanation of your results.

```python
import netCDF4 as nc
import numpy as np
import matplotlib.pyplot as plt

# Open ERA-Interim dataset
file = '/data/era-interim/era_interim_monthly_197901_201512_upscaled.nc_ann'

ds = nc.Dataset(file)

# Inspect dataset
print(ds)

# Extract variables
t2m = ds.variables['t2m'][:]
lat = ds.variables['lat'][:]
lon = ds.variables['lon'][:]

print(t2m.shape)
print(lat.shape)
print(lon.shape)

# Earth's radius (meters)
R = 6371000

# Grid resolution (degrees)
dlat = abs(lat[1] - lat[0])
dlon = abs(lon[1] - lon[0])

# Convert longitude resolution to radians
dlon_rad = np.deg2rad(dlon)

# Calculate latitude boundaries
lat_lower = lat - dlat / 2
lat_upper = lat + dlat / 2

# Ensure boundaries don't exceed the poles
lat_lower = np.clip(lat_lower, -90, 90)
lat_upper = np.clip(lat_upper, -90, 90)

# Convert latitude boundaries to radians
lat_lower_rad = np.deg2rad(lat_lower)
lat_upper_rad = np.deg2rad(lat_upper)

# Calculate area of each latitude band (per grid cell)
area_lat = R**2 * dlon_rad * (
    np.sin(lat_upper_rad) - np.sin(lat_lower_rad)
)

# Construct 2D array of pixel areas
area = np.repeat(area_lat[:, np.newaxis], len(lon), axis=1)

print(area.shape)

# Naive spatial mean
naive_mean = np.ma.mean(t2m, axis=(1, 2))

# Calculate area-weighted mean for each year
weighted_mean = []

for year in range(t2m.shape[0]):

    # Extract temperature for this year
    temp = t2m[year, :, :]

    # Account for masked or missing values
    valid = ~np.ma.getmaskarray(temp) & np.isfinite(temp)

    # Calculate weighted mean
    mean = np.sum(temp[valid] * area[valid]) / np.sum(area[valid])

    weighted_mean.append(mean)

weighted_mean = np.array(weighted_mean)

# Create years array
years = np.arange(1979, 2016)

# Convert Kelvin to Celsius
naive_c = naive_mean #  - 273.15
weighted_c = weighted_mean #  - 273.15

# Plot
plt.figure(figsize=(12, 6))

plt.plot(years, naive_c,
         label='Naive Mean',
         color='red')

plt.plot(years, weighted_c,
         label='Area-Weighted Mean',
         color='blue')

plt.xlabel('Year')
plt.ylabel('Global Mean Temperature (°C)')
plt.title('ERA-Interim Global Mean Temperature (1979–2015)')
plt.legend()
plt.grid(alpha=0.3)

plt.show()
```

# 4. North Carolina county level elevation statistics (20 pts)

Given the shapefile of county boundaries (`/data/HW/HW2/NCDOT_County_Boundaries.shp`) and the USGS National Elevation Dataset (`/data/NED/NED.vrt`), compute the following statistics: the spatial mean, the spatial standard deviation, and the coefficient of variation of elevation per NC county. Once this database is assembled, use bar plots to show the results across all the counties (one bar plot per statistic). Explain the results. To make things simple you will want to use the Plate Carree projection (regular lat/lon grid) at a 0.01 by 0.01 arcdegree spatial resolution.

```python
%%bash
# Northwest
gdalwarp \
    -te -85 35.25 -80 37 \
    -tr 0.01 0.01 \
    -r near \
    -overwrite \
    /data/NED/NED.vrt \
    NED_NC_NW.tif

# Northeast
gdalwarp \
    -te -80 35.25 -75 37 \
    -tr 0.01 0.01 \
    -r near \
    -overwrite \
    /data/NED/NED.vrt \
    NED_NC_NE.tif

# Southwest
gdalwarp \
    -te -85 33.5 -80 35.25 \
    -tr 0.01 0.01 \
    -r near \
    -overwrite \
    /data/NED/NED.vrt \
    NED_NC_SW.tif

# Southeast
gdalwarp \
    -te -80 33.5 -75 35.25 \
    -tr 0.01 0.01 \
    -r near \
    -overwrite \
    /data/NED/NED.vrt \
    NED_NC_SE.tif

gdal_merge.py \
    -o NED_NC.tif \
    NED_NC_NW.tif \
    NED_NC_NE.tif \
    NED_NC_SW.tif \
    NED_NC_SE.tif

gdalinfo NED_NC.tif | grep -E "Size is|Pixel Size|Upper Left|Lower Right"
```

```python
import geopandas as gpd
import rasterio
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from rasterio.mask import mask

county_file = '/data/HW/HW2/NCDOT_County_Boundaries.shp'

counties = gpd.read_file(county_file)

# Convert to same CRS as NED raster
counties = counties.to_crs(epsg=4326)

print(counties.head())

results = []

with rasterio.open('NED_NC.tif') as src:
    print(src.crs)
    print(src.res)
    for index, county in counties.iterrows():

        # County name
        county_name = county['CountyName']

        # Extract raster cells within county
        elevation, transform = mask(
            src,
            [county.geometry],
            crop=True,
            filled=False
        )

        # First (and only) raster band
        elevation = elevation[0]

        # Remove masked/NoData cells
        values = elevation.compressed()

        # Calculate statistics
        mean_elev = np.mean(values)
        std_elev = np.std(values)
        cv_elev = std_elev / mean_elev

        # Save results
        results.append({
            'County': county_name,
            'Mean': mean_elev,
            'Std': std_elev,
            'CV': cv_elev
        })


df = pd.DataFrame(results)

df = df.sort_values('County')

print(df)
print(len(df))

plt.figure(figsize=(20, 6))

plt.bar(df['County'], df['Mean'])

plt.xlabel('County')
plt.ylabel('Mean Elevation (m)')
plt.title('Mean Elevation by North Carolina County')

plt.xticks(rotation=90, fontsize=7)

plt.tight_layout()
plt.show()

plt.figure(figsize=(20, 6))

plt.bar(df['County'], df['Std'])

plt.xlabel('County')
plt.ylabel('Standard Deviation of Elevation (m)')
plt.title('Elevation Standard Deviation by North Carolina County')

plt.xticks(rotation=90, fontsize=7)

plt.tight_layout()
plt.show()

plt.figure(figsize=(20, 6))

plt.bar(df['County'], df['CV'])

plt.xlabel('County')
plt.ylabel('Coefficient of Variation')
plt.title('Elevation Coefficient of Variation by North Carolina County')

plt.xticks(rotation=90, fontsize=7)

plt.tight_layout()
plt.show()
```

# 5. Flying from LAX to LHR (20 pts)

Create a rotating globe that tracks an actual flight from Los Angeles to London that happened between 2020/09/17-2020/09/18. The file `/data/HW/HW2/lax2lhr_20200917.pck` contains the GPS location of the flight at discrete timestamps during the flight. The goal here is to create a movie where each frame builds on the previous one to build out the time evolving flight path. The flight path should be drawn as lines joined at nodes. Each frame update should correspond to a new timestamp. Furthermore, for each timestamp, the globe should be centered on the timestamp's corresponding longitude. Finally, add the time evolving rainfall maps that was forecasted during each timestamp. For this part, you will use a 9/17/2020 forecast from NOAA's Global Forecasting System (GFS) which you can find in  `/data/HW/HW2/gfs20200917_gfs_0p25_1hr_00z.nc`. The variable of interest in that dataset will be `pratesfc`. Note that there are many ways to do this problem so feel free to explore your own ideas. However, for reference, below is an example of what the last frame might look like. We look forward to seeing what you come up with! 


```python
import pickle
import numpy as np
import pandas as pd
import xarray as xr
import matplotlib.pyplot as plt
import cartopy.crs as ccrs
import cartopy.feature as cfeature
from matplotlib.animation import FuncAnimation
from IPython.display import HTML

# assign flight data filepath to a variable
file = '/data/HW/HW2/lax2lhr_20200917.pck'
# open the flight data file
with open (file, 'rb') as f:
    flight_data = pickle.load(f)
# convert to pandas dataframe
flight_data = pd.DataFrame(flight_data)

# forecasting data
gfs_file = '/data/HW/HW2/gfs20200917_gfs_0p25_1hr_00z.nc'
gfs_data = xr.open_dataset(gfs_file)


# convert flight dates to datetime values 
flight_data['date'] = pd.to_datetime(flight_data['date'])


gfs_lats = gfs_data['lat'].values
gfs_lons = gfs_data['lon'].values
# Convert GFS longitudes from 0-360 to -180-180 if necessary
gfs_lons = ((gfs_lons + 180) % 360) - 180
# Sort longitude after conversion
lon_order = np.argsort(gfs_lons)
gfs_lons = gfs_lons[lon_order]

# create the animation
fig = plt.figure(figsize=(12, 8))

def update(frame):

    # Clear the figure so that we can create a new projection
    # centered on the current aircraft longitude.
    fig.clear()

    current_lon = flight_data['lon'].values[frame]
    current_lat = flight_data['lat'].values[frame]
    current_time = flight_data['date'].iloc[frame]

    # --------------------------------------------------------
    # Create globe centered on aircraft longitude
    # --------------------------------------------------------

    ax = fig.add_subplot(
        1, 1, 1,
        projection=ccrs.Orthographic(
            central_longitude=current_lon,
            central_latitude=30
        )
    )

    ax.set_global()

    ax.add_feature(cfeature.LAND, facecolor='lightgray')
    ax.add_feature(cfeature.OCEAN, facecolor='white')
    ax.add_feature(cfeature.COASTLINE)
    ax.add_feature(cfeature.BORDERS, linewidth=0.4)

    # --------------------------------------------------------
    # Find GFS forecast closest to aircraft timestamp
    # --------------------------------------------------------

    gfs_time = pd.to_datetime(gfs_data['time'].values)

    time_index = np.argmin(
        np.abs(gfs_time - current_time)
    )

    rain = gfs_data['pratesfc'].isel(time=time_index).values

    # Reorder longitude dimension to match converted longitudes
    rain = rain[:, lon_order]

    # --------------------------------------------------------
    # Plot precipitation
    # --------------------------------------------------------

    ax.contourf(
        gfs_lons,
        gfs_lats,
        rain,
        levels=20,
        transform=ccrs.PlateCarree(),
        cmap='Blues',
        alpha=0.7
    )

    # --------------------------------------------------------
    # Draw flight path UP TO CURRENT TIMESTAMP
    # --------------------------------------------------------

    ax.plot(
        flight_data['lon'].values[:frame + 1],
        flight_data['lat'].values[:frame + 1],
        '-o',
        transform=ccrs.Geodetic(),
        linewidth=2,
        markersize=3,
        color='red'
    )

    # Highlight current aircraft position
    ax.plot(
        current_lon,
        current_lat,
        'o',
        transform=ccrs.PlateCarree(),
        markersize=8,
        color='black'
    )

    # --------------------------------------------------------
    # Title
    # --------------------------------------------------------

    ax.set_title(
        'LAX → LHR\n'
        + current_time.strftime('%Y-%m-%d %H:%M UTC'),
        fontsize=14
    )


animation = FuncAnimation(
    fig,
    update,
    frames=len(flight_data),
    interval=200
)

# Prevent Jupyter from displaying a separate static figure
plt.close(fig)

# Display animation in the notebook
HTML(animation.to_jshtml())
```
